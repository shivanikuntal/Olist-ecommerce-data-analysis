# **DATA_CLEANING**

```python
Orders['order_status'].value_counts()
```

```python
pd.crosstab(
      Orders['order_status'],
          Orders['order_delivered_customer_date'].isnull()
          )
```

```python
Orders[
      (Orders['order_status'] == 'delivered') &
          (Orders['order_delivered_customer_date'].isnull())
          ]
```

```python
Orders[
      (Orders['order_status'] == 'delivered') &
          (Orders['order_delivered_customer_date'].isnull())
          ].T
```

```python
order_date_cols = [
      'order_purchase_timestamp',
          'order_approved_at',
              'order_delivered_carrier_date',
                  'order_delivered_customer_date',
                      'order_estimated_delivery_date'
                      ]

for col in order_date_cols:
       Orders[col] = pd.to_datetime(Orders[col], errors='coerce')


Orders_pay['payment_value'] = pd.to_numeric(
Orders_pay['payment_value'],
errors='coerce'
)

Orders_item['shipping_limit_date'] = pd.to_datetime(
Orders_item['shipping_limit_date'],
errors='coerce'
)
```

```python
review_date_cols = [
      'review_creation_date',
          'review_answer_timestamp'
          ]

for col in review_date_cols:
         Review[col] = pd.to_datetime(
         Review[col],
         errors='coerce'
      )
```

```python
Orders.drop_duplicates(inplace=True)
Customers.drop_duplicates(inplace=True)
Orders_item.drop_duplicates(inplace=True)
Orders_pay.drop_duplicates(inplace=True)
Products.drop_duplicates(inplace=True)
Review.drop_duplicates(inplace=True)
Sellers.drop_duplicates(inplace=True)
Geolocation.drop_duplicates(inplace=True)
Category_translation.drop_duplicates(inplace=True)
```

```python
Products['product_category_name'] = (
      Products['product_category_name']
          .fillna('unknown')
          )

numeric_product_cols = [
'product_name_lenght',
'product_description_lenght',
'product_photos_qty',
'product_weight_g',
'product_length_cm',
'product_height_cm',
'product_width_cm'
]

for col in numeric_product_cols:
     Products[col] = Products[col].fillna(
     Products[col].median()
     )
```

```python
Review['review_comment_title'] = (
Review['review_comment_title'].fillna('')
      )

Review['review_comment_message'] = (
Review['review_comment_message'].fillna('')
      )
```

```python
Customers['customer_city'] = Customers['customer_city'].str.strip().str.lower()
Customers['customer_state'] = Customers['customer_state'].str.strip().str.upper()

Sellers['seller_city'] = Sellers['seller_city'].str.strip().str.lower()
Sellers['seller_state'] = Sellers['seller_state'].str.strip().str.upper()

Orders['order_status'] = Orders['order_status'].str.strip().str.lower()

Orders_pay['payment_type'] = Orders_pay['payment_type'].str.strip().str.lower()
```

```python
Orders['delivery_days'] = (
Orders['order_delivered_customer_date']
- Orders['order_purchase_timestamp']
).dt.total_seconds() / 86400
```

```python
Orders['delivery_delay_days'] = (
Orders['order_delivered_customer_date']
  - Orders['order_estimated_delivery_date']
   ).dt.total_seconds() / 86400
```

```python
Orders['is_late'] = (
      Orders['delivery_delay_days'] > 0)

```

```python
Orders['order_month'] = (
      Orders['order_purchase_timestamp']
          .dt.to_period('M')
              .astype(str)
              )
```

```python
Orders.to_csv('clean_Orders.csv', index=False)

Customers.to_csv('clean_Customers.csv', index=False)

Orders_item.to_csv('clean_Order_Items.csv', index=False)

Orders_pay.to_csv('clean_Payments.csv', index=False)

Products.to_csv('clean_Products.csv', index=False)

Review.to_csv('clean_Reviews.csv', index=False)

Sellers.to_csv('clean_Sellers.csv', index=False)

Geolocation.to_csv('clean_Geolocation.csv', index=False)

Category_translation.to_csv('clean_Category_Translation.csv', index=False)
```
